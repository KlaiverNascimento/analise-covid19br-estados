```python
url_covid = pd.read_csv('../fontes/cases-brazil-states.csv', sep = ',')
url_covid 
```

```python
df_tratado = url_covid [['date', 'country', 'state', 'newDeaths', 'deaths',	'newCases', 'totalCases','deaths_per_100k_inhabitants'	]]

df_tratado = df_tratado.rename(columns ={
    'date': 'DAT_OCORRIDO',
    'country': 'NME_REGIAO_OCORRIDO', 
    'state': 'NME_ESTADO_OCORRIDO',
    'newDeaths': 'NUM_NOVO_OBITO', 
    'deaths': 'NUM_OBITO',
    'newCases': 'NUM_NOVOS_CASOS', 
    'totalCases': 'NUM_TOTAL_CASOS',
    'deaths_per_100k_inhabitants': 'NUM_OBITOS_100K_HABITANTES'
})
df_tratado
```

Daqui para baixo, achei necessário adicionar os devidos tratamentos me baseando nas colunas que eu já havia escolhido para prencheer dentro do banco desse projeto 

```python
df_tratado = df_tratado[df_tratado['NME_ESTADO_OCORRIDO'] != 'TOTAL'].reset_index(drop = True) #Aqui, dropei quando era "Verdadeiro" dentro de campos NME_ESTADO_OCORRIDO linhas 
#que continham o valor prenchido apenas com "TOTAL" aonde deveria aparecer os estados brasileiros. 
df_tratado.describe() ## Utilizei esse describe para verificar principalmente dentro das colunas NUM_OBITO, NUM_NOVO_OBITO, NUM_NOVOS_CASOS, NUM_TOTAL_CASOS a evidencia de valores nulos
#Já que durante a pandemia o estados corrigia um erro de contagem de dias anteriores, esses campos podiam ficar negativos. 
```

Efetuando a conexão com o banco de dados do SQLITE3 para subir e criar a tabela ODS_COVID_ESTADOS

```python
dados_db = "../database/covid19br.db"
conn = sqlite3.connect(dados_db)
```

```python
df_tratado.to_sql(
    name= "ODS_COVID_ESTADOS",
    con= conn, 
    if_exists= "replace",
    index=False, 
)
```
